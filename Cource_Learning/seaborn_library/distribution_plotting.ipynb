# distribution plots

> 1. Use for Univariate Analysis - analysis on single column
> 2. Use to find out the distribution - ranges, mean, mode, central tendency, min, max, avg
> 3. to check data has outlier or not?
> 4. Data is binoial or not ?

>> PLots under distribution plotting:
>   1. Hist plot
>   2. Kde plot
>   3. Rug plot

>> Functions: 
>   1. Figure level function: distplot()
>   2. Axis level function:  
>       1. histplot()
>       2. kdeplot()
>       3. rugplot()

## KDE PLot : Kernel Density Estimation

>   Rather that using descrite bins, KDE plot smooth the observation with Gauscian kernel, producing continuous density estimation

```python
tips.head()
```

```python
sns.kdeplot(data=tips, x='total_bill')
```

```python
# kde plot represent the density of probability of total_bill 20 is 5% i.e 0.05
# from this we can see range of distribution from start to end like total_bill from 50-60 is nearly 0.3-4%
```

```python
sns.histplot(data=tips, x='total_bill') 
```

```python
# kdeplot using figure function
sns.displot(data=tips, x='total_bill', kind='kde')
```

```python
# probability of density function of total_bill with compare to Sex Compare Male vs Female within each panel (Color-wise)
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex', fill=True)
```

> How does the distribution of total bill amounts differ by gender, and how does that comparison change for smokers vs non-smokers?”

```python
# probability of density function of total_bill with compare to Sex 
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex', fill=True,col='smoker')  
```

# Rug Plot

> Plot marginal distribution by drawing ticks along with the x and y axis
> This function intended to complement other plots by showing location of individual observation in an unobstructive way

```python
sns.kdeplot(data=tips, x='total_bill')
sns.rugplot(data=tips, x='total_bill')
```

```python
sns.histplot(data=tips, x='total_bill', y='tip')
```
