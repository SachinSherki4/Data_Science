### One-Hot Incoding for Normalisation and Standardisation - to convert categorical data into Numerical data

1. Pandas df provide one function - pd.get_dummies(data, column)
2. Here in One-Hot incoding - every unique data get one idividual column so n unique values in categorical column we get n unique coumns with their name
3. For Algorithm, feading numerical values is much more easier and simpler as compare to feading categorical data so for this purpose, we convert categorical data into numerical data
4. True -mean 1 and False means 0

```python
new_df = pd.get_dummies(titanic_data, columns=['Sex','Embarked']) # Sex_female & Sex_male as true/false representation while your original data is secure
```

```python
new_df # Embarked_C/Q/S Sex_Female/Male
```

### Import LabelIncoder package from sklearn.preprocessing
