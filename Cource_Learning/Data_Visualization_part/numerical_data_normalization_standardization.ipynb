### Scikit-learn library for further processing data - scalling technique
Steps 
1. first import the required package
2. Initialise the method (MinMaxScaller()) - using that variable can call all available functions
3. Now using fit_transform(dataframe) - convert all data into standard values (0-1)

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

### MinMaxScaller - converting all data set into standard 0-1 format

```python
#Initialise the method (MinMaxScaller()) - using that variable can call all available functions
min_max_scaller=MinMaxScaler() # using this variable can call all functions
```

```python
#Now using fit_transform(dataframe) - convert all data into standard values (0-1)
salary_data_transform= min_max_scaller.fit_transform(salary_data)
```

```python
salary_data_transform # as this is in nd.array format which we need to convert into DataFrame
```

```python
new_salary_data=pd.DataFrame(salary_data_transform, columns=['YearsExperience','Salary'])
```

```python
new_salary_data
```

```python
plt.scatter(x='YearsExperience',y='Salary',data=new_salary_data)
```

### StandardScaler() - based on 3sigma, 6sigma - Standard Normal Distribution - using ZScore
ZScore - represent how far value from mean in terms of Standard Distribution 
1. ZScore = (x - mean)/ sigma(S.D)
2. ZScore use to copare different type of data o the same scale

```python
std_scaler= StandardScaler()
```

```python
transform_sal_data=std_scaler.fit_transform(salary_data)
```

```python
transform_sal_data
```

```python
new_std_salary_data=pd.DataFrame(transform_sal_data, columns=['YearsExperience','Salary'])
```

```python
new_std_salary_data.head()
```

```python
plt.scatter(x='YearsExperience',y='Salary',data=new_std_salary_data)
```

```python
# correlation of these data
new_std_salary_data['YearsExperience'].corr(new_std_salary_data['Salary']) # 0.9782416184887599 - very high 
```
