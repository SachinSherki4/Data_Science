### 1. Create a new column ROI (Return on Investment) = (gross - budget) / budget.

```python
movie_df['budget'].isna().sum()
```

```python
movie_df['gross'].isna().sum()
```

```python
movie_df['gross'].fillna(value=movie_df['gross'].mean(), inplace=True)
```

```python
movie_df['gross'].isna().sum()
```

### here if calculate ROI directly without removing outlier value will be very distorted

```python
movie_df['ROI']=(movie_df['gross']- movie_df['budget'])/movie_df['budget']
```

```python
movie_df['ROI'].sort_values(ascending=False)
```

```python
movie_df['gross'].describe()
```

```python
movie_df['ROI'].describe()
```

```python
def outlier_removal(col):
    q1, q3=np.percentile(col, [25,75])
    IQR = q3- q1
    lower_range= q1 - (1.5 * IQR)
    higher_range = q3 + (1.5 * IQR)
    return lower_range, higher_range
```

```python
lower, upper= outlier_removal(movie_df['gross'])
```

```python
movie_df=movie_df[(movie_df['gross'] > upper) | (movie_df['gross'] < lower)]
```
