```python
import pandas as pd
import numpy as np 
```

```python
sample_data= {'Date': ['2025-03-22','2025-03-23','2025-03-24']}
sample_df=pd.DataFrame(data=sample_data)
```

```python
sample_df
```

```python
sample_df.info() #object - considering as categorical variable
```

We have date columns but Date is not consider as Date instead it is considering as object-categorical data

```python
sample_df['converted_datetime']=pd.to_datetime(sample_df['Date'])
```

```python
sample_df.info() #datetime64[ns]
```

Here if we have DateTime object variable we can easily separate variable based on date/time/year

Formatting DATe

```python
sample_data2= {'Date': ['22-03-2025','23-03-2025','24-03-2025']}
sample_df2=pd.DataFrame(data=sample_data2)
```

```python
sample_df2.info()
```

```python
sample_df2
```

```python
sample_df2['converted_datetime']=pd.to_datetime(sample_df2['Date'], format='%d-%m-%Y')
```

```python
sample_df2
```

Extracting specific data to create particular columns

```python
sample_df2['Month']= sample_df2['converted_datetime'].dt.month
```

```python
sample_df2['Day']=sample_df2['converted_datetime'].dt.day
```

```python
sample_df2['Year']=sample_df2['converted_datetime'].dt.year
```
