## Scatter plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# creating random range
x= np.linspace(-10, 10, 50)
# creating random y number
y=  10*x + 3 + np.random.randint(0,300, 50)
y
```

```python
plt.scatter(x,y)
```

```python
df= pd.read_csv('../../Data_Files/IPL_data/batter.csv')
df= df.head(50)
df
```

```python
plt.scatter(df['avg'], df['strike_rate'])
plt.title("Avg and Strike Rate Analysis")
plt.xlabel("Average")
plt.ylabel("Strike Rate")
```

```python
tips= sns.load_dataset('tips')
tips
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
plt.xlabel("total_bill")
plt.ylabel("tips")
plt.title("total_bill vs tips")
```

```python
# we can add third columns as parameter to represent scatter graph 
plt.scatter(tips['total_bill'],tips['tip'], s=tips['size']*20)
plt.xlabel("total_bill")
plt.ylabel("tips")
plt.title("total_bill vs tips vs Size")
```

```python
# scattering using plt.plot() - faster as compare to scatter which is slower
plt.plot(tips['total_bill'], tips['tip'])
```
