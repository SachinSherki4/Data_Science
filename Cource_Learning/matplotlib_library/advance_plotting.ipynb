```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Showing different color with dfferent categories

```python
iris= pd.read_csv('../../Data_Files/IPL_data/iris.csv')
iris
```

```python
# here we have 3 different species of flowers - here we cant bifercate which flower are dfferent
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'])

plt.colorbar() # to represent the color map of different species
```

```python
# we can assign an number or sequence to all species and show their plot
iris['Species'].value_counts()
```

```python
iris['Species']= iris['Species'].replace({'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica':2})
iris.sample(5)
```

```python
# can see different flower species c=iris['Species']
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet') # cpam= to show color variance
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.title("Flower SepalLengthCm vs PetalLengthCm")
plt.colorbar() # to represent the color map of different species
```

### Plot Size

```python
plt.figure(figsize=(7,4))
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet') # cpam= to show color variance
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.title("Flower SepalLengthCm vs PetalLengthCm")
```

### Anotations

```python
batter= pd.read_csv('../../Data_Files/IPL_data/batter.csv')
batter= batter.head(100).sample(25, random_state=5) 
batter # top 100 and selecting random 25/100 
```

```python
plt.scatter(batter['avg'], batter['strike_rate'])
```

```python
# we cant identify which point represent which batter 
# to show name of batter for each batsman we can use below code plt.text(x, y, 'text')

plt.figure(figsize=(14,8))
plt.scatter(batter['avg'], batter['strike_rate']) # this will draw graph
plt.grid()
plt.title("Batsman Avg vs Strike Rate")
plt.xlabel("Average")
plt.ylabel('Strike Rate')
# this will assign names to each batter plt.text(x, y, text, fontdict={'size':12, 'color': 'red})
for i in range(batter.shape[0]):
    plt.text(batter['avg'].values[i], batter['strike_rate'].values[i], batter['batter'].values[i],fontdict={'size':12, 'color': 'red'})
```

### Horizontal and Vertical Line graph

```python
# if want to draw a line in horizontal and vertical ploting different strike rate and avg
plt.figure(figsize=(12,8))
plt.scatter(batter['avg'], batter['strike_rate']) # this will draw graph
#plt.grid()
plt.title("Batsman Avg vs Strike Rate")
plt.xlabel("Average")
plt.ylabel('Strike Rate')

# draw horizontal line - batter wit strike rate 130 and above
plt.axhline(130, color= 'red')
plt.axhline(140, color= 'green')
# draw vertical line for batter witht avegare 30 and more
plt.axvline(30, color='red')
plt.axvline(35, color='green')

# this will assign names to each batter plt.text(x, y, text, fontdict={'size':12, 'color': 'red})
for i in range(batter.shape[0]):
    plt.text(batter['avg'].values[i], batter['strike_rate'].values[i], batter['batter'].values[i])
```

### PLottng multiple graphs

```python
batters= pd.read_csv('../../Data_Files/IPL_data/batter.csv')
batters
```

```python
plt.figure(figsize=(15,8))
plt.scatter(batters['avg'], batters['strike_rate'])
```

```python
# using plt.subplots() - we can plot mutplie graphs
# fig, ax= plt.subplots(nrows=2, ncols=1, figsize=(15,8)) = two graphs 
# fig = represent 2 graphs\
# ax = represent two objects

fig, ax= plt.subplots(nrows=2, ncols=1) 
```

```python
fig
```

```python
ax
```

```python
fig, ax= plt.subplots(figsize=(12,8)) # this will create default 1 graph

ax.scatter(batters['avg'], batters['strike_rate'], color='red',marker='+')
ax.set_title("Batsman Average vs Strike Rate")
ax.set_xlabel("Average")
ax.set_ylabel("Strike Rate")
```

```python
fig, ax= plt.subplots(nrows=2, ncols=1, figsize=(12,8),sharex=True)

ax[0].scatter(batters['avg'], batters['strike_rate'], color='red',marker='+', label="Avg vs Strike Rate")
ax[1].scatter(batters['avg'], batters['runs'], color='blue', label='Avg vs Runs')

ax[0].set_title("Avg vs Strike Rate")
ax[0].set_xlabel('Avg')
ax[0].set_ylabel('Strike Rate')
ax[1].set_title("Avg vs Runs")
ax[1].set_xlabel('Avg')
ax[1].set_ylabel('Runs')
plt.legend()
```

### plotting 4 graphs

```python
fig, ax= plt.subplots(nrows=2, ncols=2, figsize=(12,8))

ax[0,0].scatter(batters['avg'], batters['strike_rate'], label="Avg vs Strike Rate", color="red")
ax[0,1].scatter(batters['avg'], batters['runs'], color='blue', label='Avg vs Runs')
ax[1,0].hist(batters['avg'], label='Average')
ax[1,1].hist(batters['strike_rate'], label='strike_rate')


ax[0,0].set_title("Avg vs Strike Rate")
ax[0,1].set_title("Avg vs Runs")
ax[1,0].set_title("Avg")
ax[1,1].set_title("Strike Rate")
```
