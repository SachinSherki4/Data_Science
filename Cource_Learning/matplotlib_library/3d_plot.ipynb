```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 3D Scatter plot

```python
# plt.subplot(projection='3d')

batter= pd.read_csv('../../Data_Files/IPL_data/batter.csv')
batter
```

```python
fig= plt.figure(figsize=(12,8))

ax= plt.subplot(projection='3d')

ax.scatter3D(batter['runs'], batter['avg'], batter['strike_rate'])
ax.set_xlabel("runs")
ax.set_ylabel("avg")
ax.set_zlabel("strike_rate")
ax.set_title("3D Scatte plot - Runs vs Avg vs Strike Rate")
```

### 3D Line Plot

```python
x=[2,5,7,23,76]
y=[1,2,0,3,12]
z=[4,0,2,56,44]

fig= plt.figure(figsize=(12,8))

ax= plt.subplot(projection='3d')
ax.scatter3D(x,y,z, s=[100,100,100,100,100])
ax.plot3D(x,y,z, color='red')

for i in range(len(x)):
    ax.text3D(x[i],y[i],z[i],s=[x[i],y[i],z[i]])
```

### Plot Surface

```python
x= np.linspace(-10,10,100)
y= np.linspace(-10,10,100)
xx, yy= np.meshgrid(x,y)
xx.shape
```

```python
yy
```

```python
# z= x**2 + y**2
zz= xx**2 + yy**2
zz.shape
```

```python
fig= plt.figure(figsize=(12,8))

ax= plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz, cmap='viridis')
fig.colorbar(p) 
```

```python
# equation z= sin(xx) + cos(yy)
zz= np.sin(xx) + np.cos(yy)
fig= plt.figure(figsize=(12,8))

ax= plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz, cmap='viridis')

fig.colorbar(p) 
```

```python
# equation yy= xx**2 - yy**2
yy= xx**2 - yy**2
fig= plt.figure(figsize=(12,8))

ax= plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,zz, cmap='viridis')
fig.colorbar(p) 
```

### Contour PLot - to show 3D graph in 2D view

```python
fig= plt.figure(figsize=(12,8))
zz= xx**2 + yy**2
ax= plt.subplot()
p=ax.contour(xx,yy,zz, cmap='viridis')
fig.colorbar(p) 
```

```python
fig= plt.figure(figsize=(12,8))
zz= xx**2 + yy**2
ax= plt.subplot()
p=ax.contourf(xx,yy,zz, cmap='viridis')
fig.colorbar(p) 
```

```python
# equation z= sin(xx) + cos(yy)
zz= np.sin(xx) + np.cos(yy)
fig= plt.figure(figsize=(12,8))

ax= plt.subplot()
p=ax.contourf(xx,yy,zz, cmap='viridis')

fig.colorbar(p) 
```
