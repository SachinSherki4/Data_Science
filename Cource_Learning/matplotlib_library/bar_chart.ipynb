```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
children=[10,30,20,40,10] # numeric
colors=['red','white','green','silver','pink'] # categorical

# vertical bar chart
# plt.bar(categorical, numeric)
plt.bar(colors, children)
plt.title("Color vs Children")
```

```python
# if want to show graph in horizontal bar chart
# when no of categoris increase it is better to choose horizontal bar chart
plt.barh(colors, children, color='black')
plt.grid()
```

```python
## plotting multiple columns side by side
df=pd.read_csv('../../Data_Files/IPL_data/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'], width=0.3)
plt.bar(df['batsman'],df['2016'], width=0.3)
plt.bar(df['batsman'],df['2017'], width=0.3)
# it is overlapping the all charts
```

```python
plt.bar(np.arange(df.shape[0])-0.2 , df['2015'],width=0.2, color='blue', label='2015')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, color='red', label='2016')
plt.bar(np.arange(df.shape[0]) +0.2, df['2017'], width=0.2, color='black', label='2017')

plt.xticks(np.arange(df.shape[0]), df['batsman']) # to show the batsman names
plt.title
plt.legend()
plt.grid()
plt.show()
```

```python
# writing x bar categories in vertical or in angle
plt.bar(df['batsman'], df['2015'])
plt.xticks(rotation='vertical')
plt.show()
```

```python
# if want to show all sum of score in same chart
plt.bar(df['batsman'],df['2017'], label='2017')
plt.bar(df['batsman'],df['2016'], bottom=df['2017'], label='2016')
plt.bar(df['batsman'],df['2015'], bottom=df['2016']+df['2017'], label='2015')
#plt.grid()
plt.legend()
plt.show()
```
