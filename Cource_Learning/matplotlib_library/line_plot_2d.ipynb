### 2D line plot

```python
# IPL data 
batsman= pd.read_csv('../../Data_Files/IPL_data/sharma-kohli.csv')
batsman.info()
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# both side by side 
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# label title
# both side by side 
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.plot(batsman['index'],batsman['V Kohli'])
plt.title('R Sharma vs V Kohli Career Comparison')
plt.xlabel("Season")
plt.ylabel('Runs Score')
```

```python
# customise color(hex) line(width and style) and marker size
plt.plot(batsman['index'],batsman['RG Sharma'], color='green', linestyle='solid', linewidth=1, marker='D') # dotted, dashdot, solid,
plt.plot(batsman['index'],batsman['V Kohli'], color= '#5E332A', linestyle='dotted', linewidth=2, marker='+', markersize=3) # marker= o, +, D, -
plt.title('R Sharma vs V Kohli Career Comparison')
plt.xlabel("Season")
plt.ylabel('Runs Score')
```

```python
# if want to print label on graph we need to call legend()
plt.plot(batsman['index'],batsman['RG Sharma'], color='green', linestyle='solid', linewidth=1, marker='D', label='R Sharma') # dotted, dashdot, solid,
plt.plot(batsman['index'],batsman['V Kohli'], color= '#5E332A', linestyle='dotted', linewidth=1, marker='+', markersize=1, label='V Kohli') # marker= o, +, D, -
plt.title('R Sharma vs V Kohli Career Comparison')
plt.xlabel("Season")
plt.ylabel('Runs Score')
plt.legend()
```
