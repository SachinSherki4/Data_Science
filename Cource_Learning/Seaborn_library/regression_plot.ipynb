```python
import pandas as pd 
import seaborn as sns 
import plotly.express as px
```

```python
tips= sns.load_dataset('tips')

tips
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

### reg plot

```python
# reg plot is axis level function
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
# the blue line represent regression line - closest line passing from all point with 95% Confidence interval
# shaded region represent marginal error 
# hue parameter is not available
```

### lmplot

```python
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

### residplot - to show error made in regression chart arount zero regression line

```python
sns.residplot(data=tips, x='total_bill', y='tip')
```
