### `Categorical Scatter Plot`

```python
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
# if want to plot scatter plot on categorical data - make sure one column must be numerical
```

### Strip plot

```python
sns.stripplot(data=tips,x='day', y='total_bill', jitter=False)
```

```python
# figure level function
sns.catplot(data=tips,x='day', y='total_bill', kind='strip', jitter=0.1, hue='sex')
```

### swarm plot - it show distribution in very good way

```python
sns.swarmplot(data=tips,x='day', y='total_bill')
```

```python
# figure level function 
sns.catplot(data=tips,x='day', y='total_bill', kind='swarm', hue='sex')
```

### Box Plot -

### Violin Plot - combination of box plot and kde plot

```python
 # it show data distribution 
sns.violinplot(data=tips, x='total_bill',y='day')
```

```python
sns.catplot(data=tips, x='total_bill',y='day', kind='violin',hue='sex', split=True)
```

### `Categorical Estimated Plot -> central tendencies`
> 1. Bar plot
> 2. Point plot
> 3. Count plot

### 1. Bar Plot - for numerical column(y)

```python
# lets plot an bar plot for male/female average total bills
sns.barplot(data=tips, x='sex',y='total_bill', hue='smoker',errorbar=None)
# average male bill= 20, female bill=18 
```

```python
# bar plot by default calculate average of numerical column- we can change it and count other parameters also like std_dev, mean, median
sns.barplot(data=tips, x='sex', y='total_bill', estimator=np.std, hue='smoker',errorbar=None) # standerd deviation of total bills
```

```python
sns.barplot(data=tips, x='sex', y='total_bill', estimator=np.median, hue='smoker',errorbar=None) # median of total_bill
```

```python
sns.barplot(data=tips, x='sex', y='total_bill', estimator=np.min, hue='smoker', errorbar=None) # minimum data
```

### Point Plot = to plot the difference between two categories

```python
sns.pointplot(data=tips, x='sex', y='total_bill',  hue='smoker',errorbar=None) # average 
```

```python
sns.pointplot(data=tips, x='sex', y='total_bill', estimator=np.mean,errorbar=None)
```

```python
sns.pointplot(data=tips, x='day', y='total_bill',hue='sex',errorbar=None) 
```

### Count Plot - usefull for categorical values

```python
sns.countplot(data=tips, x='sex', hue='day' )
```

```python
sns.countplot(data=tips, x='sex', hue='smoker')
```

### Facet plot

```python
### Facet plot
sns.catplot(data=tips, x='sex', y='total_bill', col='smoker', row='time',kind='strip')
```
