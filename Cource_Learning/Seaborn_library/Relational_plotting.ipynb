> 2. Bivariate Analysis

>   2. Line Plot

### 1. Scatter Plot

```python
import seaborn as sns 
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips= sns.load_dataset('tips')
tips 
```

```python
# scatter plot - generally plot for two Numerical columns 
# Scatter plot - is Axis level function
sns.scatterplot(data= tips,x='total_bill',y='tip')
# can see the relation is linear - means if total_bills increases -> tip amount also increases 
```

```python
# hue parameter - to get extra info from two relational grpah
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex', style='time',size='size')
```

```python
# if want Figure level functions - relplot() - little bit sqaure shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex',style='time',size='size')
```

### 2. Line PLot - x axis (time ) - time series data - line plot

```python
sns.lineplot(data=tips, x='time',y='total_bill')
```

```python
gap= px.data.gapminder()
gap
```

`PLot graph of India - Life expentency of India 1952-2007`

```python
temp_df=gap[gap['country']=='India']
temp_df
```

```python
### Axis level function
sns.lineplot(data=temp_df, x='year',y='lifeExp') 
```

```python
sns.scatterplot(data=temp_df, x='year',y='lifeExp') 
```

>   I want to plot LifeExp grpah of Pak and China side by side of India

```python
#hue > stype
temp_df=gap[gap['country'].isin(['India','China','Pakistan'])]
temp_df
```

```python
# Axis level graph
sns.lineplot(data=temp_df, x='year',y='lifeExp', hue='country')
```

```python
# Fig level graph - fig level has access for entire fig so it added legend at outside graph

sns.relplot(data=temp_df, x='year',y='lifeExp', kind='line', hue='country')
```

```python
# if want Figure level functions - relplot() - little bit sqaure shape
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line')
```

```python
temp_df=gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
# style - function allow to show extra piece of information like which country belong to which continent
sns.relplot(data=temp_df, x='year',y='lifeExp', kind='line', hue='country',style='continent', size='continent')
```

### 3. Facet Plot - can create multiple colums on particular categorical column (multi-grid)

>   1. Facet plot - work only with Figure level functions - Relplot()
>   2. It will not work with scatter and line plot axis level functions - axis level function do not have control over entire figure

```python
# using hue
sns.relplot(kind='scatter',data=tips, x='total_bill',y='tip', hue='sex') 
```

```python
# row multi-grid 2*1
sns.relplot(kind='scatter',data=tips, x='total_bill',y='tip', row='sex')
```

```python
sns.relplot(kind='scatter',data=tips, x='total_bill',y='tip', col='sex')
```

```python
sns.relplot(kind='scatter',data=tips, x='total_bill',y='tip', row='sex', col='smoker')
```

```python
sns.relplot(kind='scatter',data=tips, x='total_bill',y='tip', row='day', col='sex')
```
