```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

## Matrix Plot

### Heatmap

```python
gap= px.data.gapminder()
gap.head()
```

```python
# above table is in long data set we need to convert it into wide format
temp_df=gap.pivot(index='country', columns='year', values='lifeExp')
temp_df 
```

```python
# now show above grid into heat map
plt.figure(figsize=(10,8))
sns.heatmap(temp_df) 
```

```python
temp_europe=gap[gap['continent']=='Europe'].pivot(index='country', columns='year', values='lifeExp')
temp_europe
```

```python
# Life Expentency in European Country
sns.heatmap(temp_europe, annot=True, linewidths=0.5, cmap='viridis') 
```
