### Uniform Cummulative Distribution Functions

```python
import numpy as np
from scipy.stats import uniform
import matplotlib.pyplot as plt 
```

1. Uniform Distribution Functions calculate Prob Density Function
2. It shows what is probability of getting particular number from given range

Que1. From given no from 10 to 20 what is prob that we get number less than 15

```python
a=10
b=20
prob_less=14
z_prob= uniform.cdf(prob_less, loc=a, scale= b-a) #scale= show range of values we have between the range
z_prob #0.5) here we can get value 10, 11, 12, 13, 14 5/10=0.5
```

What is prob that is is more than 17

```python
# more than 17
prob_more= 17
z_prob= 1- uniform.cdf(prob_more, loc=a, scale=b-a)
z_prob #0.300
```

Que1. Dice rolling- 
1. what is prob rolling a number <=4
2. What is the mean and variance of the outcome?

```python
a=1 
b=6
uni_dist=uniform(loc=a, scale=b-a)
#1. what is prob rolling a number <=4

less_equal_4= uni_dist.cdf(4)
print(f"Prob of getting less_equal_4 is {less_equal_4}")

```

```python
#2. What is the mean and variance of the outcome
print(f"Mean of outcome is : {uni_dist.mean()}") #3.5
print(f"Variance : {uni_dist.var()}") #2.08333
```

Que3. A bus arrives randomly between 9:00 AM and 9:30 AM.
1. What is the probability that it arrives before 9:10?
2. What is the expected (average) arrival time?

### follow below approach while dolving such scenarios
1. First set a=start and b= range
2. convert those into dist=uniform(loc=a, scale= b-a) 
3. then find prob based on scenarios
4. for mean(), use dist.mean()

```python
# 1. What is the probability that it arrives before 9:10?
a=0
b=30
prob1=10
dist=uniform(loc=a, scale=b-a)
z_prob= dist.cdf(prob1)
z_prob #0.333
```

```python
#2. What is the expected (average) arrival time?
dist.mean() #(15.0)
print(f"Mean time of bus arrival will be 9:{int(dist.mean())}") #Mean time of bus arrival will be 9:15
```
