```python
import numpy as np
import pandas as pd
import  matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import bernoulli, binom, poisson
```

### Bernaulli Distribution
Que1. Machine has 70% working record for any particular day, what is prob that machine will work in ext 5 days

```python
# here probability of machine will work is 70% = 
p= 0.7
# as machine will work or not work [0,1] this is example of bernaulli distribution
x= [0,1]
# for discrete variable probalibility distribution - we refer it as Probability Mass Function (PMF)
pmf = bernoulli.pmf(x,p)
plt.bar(x, pmf)
```

```python
print(pmf)
```

### Binomial Distribution 
Que2. What is a probability that 3 Heads will comes if i toss coins 5 times

```python
# n = 5 as total number of trials
# k = 3 no for which to find prob
# p = 0.5 probability of outcome for the event
n=5
k=3
p=0.5
q=np.arange(11)
pmf= binom.pmf(k,n,p)
pmf # 31249
```

```python
plt.bar(q, pmf)
```

### Poisson Distribution
Que1. In an avg i recieve 3 emails in one hr, what is prob that i will recieve 5 emails in 1 hr

```python
# here mean_avg = 3 is total avg is 3 it is not exactly every time 3 email evry time, the no may be vary for no of event and their avg is 3
mean_avg= 3
n=5
pmf= poisson.pmf(n, mean_avg)
pmf #0.10081
```

```python
mean_avg= 3
n=2
pmf= poisson.pmf(n, mean_avg)
pmf # 0.2240 
```

```python
plt.bar(n, pmf)
```
