```python
from scipy.stats import bernoulli
import matplotlib.pyplot as plt
```

### Bernoulli Distribution
Model a single trial with two outcome
Formula : 
P(X=x)= p^x (1-p)1-x
p= prob 
x= finding no

Que1. Will a customer purchase a product if the probability of purchase is 0.3?

```python
p= 0.3
n=[0,1]
pmf= bernoulli.pmf(n,p)
pmf[1] #[0.3] 
# Plot distribution
plt.hist(pmf, bins=2, edgecolor="black", density=True)
plt.xticks([0,1])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title(f"Bernoulli Distribution (p={p})")
plt.show()
```
