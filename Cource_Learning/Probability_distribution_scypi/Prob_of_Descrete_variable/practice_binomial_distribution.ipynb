### Binomial Distribution

1. Binomial distribution applicable for the problem when their is repeatation of same event to find probability of event (repeated trial of same thing)

Formula : (n/k) p^k (1-p)^ n=k
where 
n= total no of trials
k= finding prob
p = given prob

Que1. A marketing campaign sends 100 emails with a 20% open rate. What’s the probability that exactly 25 people open it?

```python
n= 100 # total no of email send
p= .20 # open rate
k= 25 # finding prob

pmf= binom.pmf(k,n,p)

pmf #0.0438
```
