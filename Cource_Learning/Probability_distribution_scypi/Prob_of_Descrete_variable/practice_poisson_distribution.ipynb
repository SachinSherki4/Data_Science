### Poisson Distribution

1. MOdel a no of event occuring in fixed interval of time(space, area, time) when event happend independently
2. Formula : p(X=x)= (e^-lambda * lambda^k)/ k!
3. lambda = mean_avg
4. k = prob find no
5. e= eiler no (e= 2.718)

Que1. A hospital gets an average of 5 emergency cases per hour. What’s the probability exactly 8 come in an hour?

```python
mean_avg= 5
n= 8
pmf= poisson.pmf(n,mean_avg)
pmf
```
