# `Business & Domain Knowledge`

`0️⃣ First Truth You Must Understand (Very Important)`

```
👉 Companies do NOT care about models, code, accuracy, Python, ML.
👉 They care about MONEY, CUSTOMERS, COST, GROWTH, RISK.
A Data Scientist is hired to HELP DECISION MAKING.
Your job is to reduce uncertainty(guess work) in business decisions using data (data backed insights).
```

`1️⃣ What Is “Business” (In Very Simple Terms)?`
```
Think of a business like a shop.

Every shop wants:
1. More customers
2. Customers to buy more
3. Spend less money to earn more
4. Keep customers coming back

That’s it.
All businesses—Amazon, Google, small kirana store—do only this.
```


`2️⃣ Where Does Data Come From?`
```
Every time a customer:
- Visits a website
- Buys a product
- Clicks an ad
- Complains
- Accepts an offer

👉 DATA IS CREATED
So data is just:  A record of human behavior

```

`3️⃣ Role of a Data Scientist in Business (Big Picture)`
```
An expert Data Scientist asks:
    -❓ Why are sales dropping?
    -❓ Which customers are valuable?
    -❓ Which campaign worked?
    -❓ Where are we losing money?
    -❓ What will happen if we do X?

He uses:
-Data
-Logic
-Statistics
-Simple ML (not fancy)
```

`4️⃣ How an Expert Data Scientist Thinks (Mental Model)`
```
🧠 Question 1: What business is this?
    Retail?
    Banking?
    Advertising?
    Healthcare?
Each domain has different goals & risks.

🧠 Question 2: Who is the customer?
    Age?
    Family?
    Income?
    Country?
Different customers behave differently.

🧠 Question 3: What action does the company want to take?
    Increase sales?
    Reduce cost?
    Stop churn?
    Target ads?

🧠 Question 4: How will my analysis change a decision?
If your analysis cannot change a decision, it is useless.

```

`5️⃣ How a Data Scientist Looks at Columns (VERY IMPORTANT)`
```
    Let’s say you receive a dataset.
    🧠 First thought is NOT:
    “Let me plot histograms”
    🧠 First thought IS:

    “What does each column represent in real life?”
   

`6️⃣ Business Thinking vs Technical Thinking`
```
    ❌ Fresher Thinking
        Plot everything
        Train models
        Improve accuracy

    ✅ Expert Thinking
        Ask why
        Reduce cost
        Increase ROI
        Improve decision quality

`7️⃣ How Data Scientist Converts Business → Data Problem`
```
    Business Question:
        “Why are customers not responding to campaigns?”

    Data Scientist converts to:
        Compare responders vs non-responders
        Income difference?
        Recency difference?
        Channel difference?

    Business Question:
        “Who should we target next?”
    Data Scientist:
        Build customer segments
        Score likelihood
        Recommend top 20%

`9️⃣ Key Skill: Talking to Business People`
```
An expert never says:
    “The model accuracy is 89%”

He says:
    “Targeting top 20% customers can increase revenue by 35%”
```

`🔟 Golden Rules You Must Remember`
```
    1️⃣ Data Scientist is a business role first
    2️⃣ Code is just a tool
    3️⃣ Every column = real human behavior
    4️⃣ If insight doesn’t change decision → useless
    5️⃣ Simple analysis > complex ML
    6️⃣ Explain like story, not math
