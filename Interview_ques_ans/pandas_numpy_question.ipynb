`Pandas — Interview Questions (15)`

`Q1. What is a DataFrame?`
> A DataFrame is a 2D table with rows and columns, like an Excel sheet.
>>  👉 Example: customer details with columns like Income, MntWines

`Q2. Difference between Series and DataFrame?`
>   1. Series = single column
> 2. DataFrame = multiple columns

```
👉 df['Income'] → Series 
👉 df[['Income','Recency']] → DataFrame 


`Q3. How do you filter rows?`
> df[df['Country'] == 'India']

`Q4. Difference between loc and iloc?`
```
loc → label based df.loc[0, 'Income']
iloc → index based  df.iloc[0, 2]
```

`Q5. How do you sort data?`
> df.sort_values('Income', ascending=False)  Used to identify top customers.

`Q6. How do you rename columns?`
>   df.rename(columns={' Income ': 'Income'})

`Q7. Why is Pandas slow for very large data?`
>   Because it works in single-machine memory, not distributed.

`Q8. How do you handle categorical columns in Pandas?`
``` 
value_counts()
groupby()
get_dummies()
```

`Q9. How do you apply a function row-wise?`
```
We can use df.apply(lambda x : expression) with lambda function to apply any filter and operation on rows
df['Age'] = df['Year_Birth'].apply(lambda x: 2025 - x)
```

`Q10. Difference between apply and vectorization?`
> apply is slower; vectorization is faster and preferred.

`Q11. When should you avoid Pandas?`
> When data is very large (GBs) → use Spark / SQL
