### Data Analysts Foundation Interview questions and answers

# `Basic Questions`

`1. What is Data Analysis?`
> Data analysis is the process of examining data to find patterns, trends, and insights that support business decisions.
>>    Marketing example: Analyzing how much customers spent on MntWines last month to understand sales performance.

`2. Why do companies need Data Analysts?`
>   To convert raw data into actionable insights that help improve revenue, efficiency, and decision-making.
>>  Finding whether customers with Teenhome = 1 spend more on MntMeatProducts.

`3. What is the role of a Data Analyst in a company?`
> A Data Analyst collects, cleans, analyzes data, and communicates insights to stakeholders.

`4. What kind of problems does a Data Analyst solve?`
>   Problems related to performance tracking, customer behavior, operational efficiency, and reporting.
>>  Why customers with high Recency (not buying recently) are declining

`5. What is descriptive analytics?`
>   Descriptive analytics explains what happened in the past using historical data.
>>  1. Total wine sales = sum of MntWines.
>>  2. Monthly sales report showing total revenue and growth percentage.

`6. What are difference types of analytics?`
>   1. `descriptive Analysis` : What is happend in past using historic data
>   2. `diagnostic Analysis` : Why something happend in past by identifying root cause
>   3. `predictive Analysis` : What is likely to happend using past trends
>   4. `prescriptive Analysis` : What action should be takenn based on insights

`7. What is diagnostic analytics?`
>   Diagnostic analytics explains why something happened by identifying root causes.
>>  Analyzing why sales dropped by checking region-wise or product-wise performance.

`7.1. When would you use diagnostic analytics?`
>   When a KPI suddenly changes and the cause is unknown.
>>  Campaign response dropped → check AcceptedCmp columns.

`8. What is predictive analytics (basic)?`
>   Predictive analytics estimates what is likely to happen using past trends.
>>  Forecasting next month’s sales using previous sales data

`9. What is prescriptive analytics (very basic)?`
>   Prescriptive analytics suggests what action should be taken based on insights.
>>  Recommending discounts in regions where sales are declining.

`10. What is a KPI?`
>   A KPI is a key performance indicator used to measure business success.
>>  Campaign response rate using Response column.

`11. Difference between metric and KPI?`
>   A metric is any measurable value; a KPI is business-critical. 
>>  1. Metric → NumWebVisitsMonth
>>  2. KPI → Response rate

`12. Data formats DA works with?`
>   CSV, Excel, databases.
>>  Marketing.csv with customer purchases.

`13. What is structured data?`
>   Tabular data with rows and columns.
>>  Customer spending columns like MntFruits.

`14. What is unstructured data?`
>   Text or images.
>>  Customer feedback comments (not in this dataset).

`15. Tools used by DA?`
>   Excel, SQL, Python, Power BI.
>>  Using Python to analyze NumCatalogPurchases.

`16. What is data cleaning?`
>   The process of fixing missing, duplicate, or incorrect data.
>>  filling missing income data, removng income outliers

`17. Why is communication important for a Data Analyst?`
>   Insights must be explained clearly to non-technical stakeholders.
>>  Explaining why married customers spend more on MntGoldProds.

# `Intermidiate Questions`

`18 How is a Data Analyst different from BI Analyst?`
>   BI focuses more on dashboards and reporting, while DA focuses on deeper analysis and insights
>>  DA finds why Campaing Response is low, BI shows response %.

`19 Difference between DA and Data Scientist?`
>   DA focuses on insights and reporting; DS focuses on predictive models and advanced analytics.
>>  DA studies campaign acceptance; DS predicts future acceptance.

`20. How does a Data Analyst support business decisions?` 
>   By providing data-backed insights that reduce guesswork.
>>  Recommending which country (Country) to target.

`21. What is trend analysis?`
>   Studying data changes over time to identify growth or decline patterns.
>>  Purchases decreasing as Recency increases.

`22. What is cohort analysis?`
>   Analyzing behavior of groups sharing common characteristics.
>>  Customers grouped by Dt_Customer year/Marital_Status.

`22.1. What is ad-hoc analysis?`
>   One-time analysis to answer a specific business question.

`22.2. What is root cause analysis?`
> Finding the primary reason behind a problem.

`25. What questions should a DA ask before analysis?`
>   1. What is the business goal?
>   2. What decision will be taken from this analysis?
>>  Should we target web or store customers?

`26. What is a dashboard?`
>   A visual summary of key metrics for monitoring performance. 
>>  Dashboard showing NumWebPurchases vs NumStorePurchases.

`27. What makes a good dashboard?`
>   Clarity, relevance, simplicity, and actionable metrics.

`28. What is data validation?`
>   Checking whether data is accurate and consistent.

`29. How do you handle missing data?`
>   Remove, replace, or flag depending on business context.
>>  1. Fill missing Income using median.
>>  2. Flag - merriage status

`30. What is outlier detection?`
>   Identifying extreme values that may distort analysis.
>>  Extreme income entries may change the mean/median values when filling missing data

`31. What is segmentation analysis?`
>   Dividing customers or data into meaningful groups.
>>  Deviding customer based on age group/ income level/marriage status/spending behaviour

`32. What is time-series data?`
>   Data collected over time intervals.
>>  Customer purchase recency.

`33. Why is business context critical?`
>   Same data can mean different things in different businesses.

`34. How do you prioritize analysis tasks?`
>   Based on business impact and urgency.
>>  Focus on campaign response first.
